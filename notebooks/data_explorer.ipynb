```python
from data_loader import read_tif, replace_invalid_and_crop, TFDatasetBuilder
import pandas as pd
import numpy as np
import ast
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('data/training_data_S2/image_metadata.csv')
```

```python
random_row = df.sample(1).iloc[0]
market_days = ast.literal_eval(random_row['market_days'])
arr = read_tif(random_row['image_file_path'])
arr = replace_invalid_and_crop(arr, 128)
vmax = np.max(arr)

rows = 3  # Calculate the number of rows needed
cols = 3  # Number of columns per row
fig, axes = plt.subplots(rows, cols, figsize=(15, 5 * rows))

# Flatten axes for easier indexing
axes = axes.flatten()

for i in range(7):
    ax = axes[i]
    ax.imshow(arr[i], cmap='gray', vmin=0, vmax=vmax)
    ax.set_title(f'Layer {i+1}\nDay Vec: {market_days[i]}')
    ax.axis('off')

# Turn off unused axes
for j in range(7, len(axes)):
    axes[j].axis('off')

plt.tight_layout()
print(vmax)
plt.show()
```

```python
builder = TFDatasetBuilder(scaling='minmax')

# Build datasets
train_ds, _, _ = builder.build_datasets(
    train_split=0.70,
    val_split=0.15,
    sample_size=100,
    batch_size=32,
)

# Retrieve one batch from the train dataset
# Take a random image from the dataset
for batch_images, batch_labels in train_ds.take(1):
    images = batch_images.numpy()
    labels = batch_labels.numpy()

random_index = np.random.randint(0, images.shape[0])
images = images[random_index:random_index+1]
labels = labels[random_index:random_index+1]

# Determine the global min and max values for all images
global_min = np.min(images)
global_max = np.max(images)

print(f"Global Min: {global_min}, Global Max: {global_max}")

rows = 3  # Calculate the number of rows needed
cols = 3  # Number of columns per row
fig, axes = plt.subplots(rows, cols, figsize=(15, 5 * rows))

# Flatten axes for easier indexing
axes = axes.flatten()

for i in range(7):
    ax = axes[i]
    ax.imshow(images[0, i, :, :, 0], cmap='gray', vmin=global_min, vmax=global_max)
    ax.set_title(f'Day {i+1} - Label: {labels[0]}')
    ax.axis('off')

# Turn off unused axes
for j in range(7, len(axes)):
    axes[j].axis('off')

plt.tight_layout()
plt.show()
```
